```python
# QUESTION 1

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

fahrenheit = (corrected_temperature * 9 / 5) + 32
print("Temperature in Fahrenheit:", fahrenheit)

readings_greater_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", readings_greater_32)

count = np.sum(corrected_temperature > 32)
print("Number of readings greater than 32°C:", count)

```

```python
# QUESTION 2

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = np.sum(steps)
print("Total steps:", total_steps)

mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)
maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum steps:", maximum)
print("Minimum steps:", minimum)

daily_total = np.sum(steps, axis=0)
print("Total steps for each day:", daily_total)

user_total = np.sum(steps, axis=1)
print("Total steps for each user:", user_total)

max_index = np.argmax(steps)

# Convert flattened index into row and column position
max_position = np.unravel_index(max_index, steps.shape)

print("\nf. Position containing maximum steps:")
print(max_position)

print("Maximum number of steps:", steps[max_position])

print("User:", max_position[0] + 1)
print("Day:", max_position[1] + 1)
```

```python
# QUESTION 3

original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]

print("Original:", original)
print("Subset:", subset)

subset[0] = 999

print("\nAfter modifying subset:")
print("Original:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()

copied_array[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)

print("\n3 x 4 Matrix:")
print(matrix)
print("\nFirst row:")
print(matrix[0, :])

print("\nLast row:")
print(matrix[-1, :])

print("\nSecond column:")
print(matrix[:, 1])

print("\nRows 1-2 and columns 2-3:")
print(matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("\nFlatten:")
print(flattened)

print("\nRavel:")
print(raveled)

raveled[0] = 100

print("\nAfter modifying ravel:")
print("Original matrix:")
print(matrix)

print("Ravel:")
print(raveled)

flattened[1] = 200

print("\nAfter modifying flatten:")
print("Original matrix:")
print(matrix)

print("Flatten:")
print(flattened)

print("\nMatrix properties:")
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python
# QUESTION 4

y = np.array([40, 65, 30, 85]) 
X = np.array([ 
[6, 70, 3], 
[5, 50, 6], 
[8, 80, 2], 
[4, 30, 8] 
]) 

print("a. Shape of X:")
print(X.shape)

print("Dimensions of X:")
print(X.ndim)

Xt = X.T
print("\nb. Transpose X.T:")
print(Xt)
print("Shape of X.T:")
print(Xt.shape)

XtX = Xt @ X
print("\nc. X.T @ X:")
print(XtX)

XtX_inverse = np.linalg.inv(XtX)
print("\nd. Inverse of X.T @ X:")
print(XtX_inverse)

Xty = Xt @ y
beta = XtX_inverse @ Xty

print("\ne. X.T @ y:")
print(Xty)

print("\nOLS coefficients (beta):")
print(beta)

print("\nf. Coefficients:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

# Sleep coefficient shows the effect of sleep hours
# on the predicted assistance score, keeping other
# features constant

# Activity coefficient shows the effect of activity
# level on the predicted assistance score, keeping
# other features constant

# Stress coefficient shows the effect of stress level
# on the predicted assistance score, keeping other
# features constant

new_user = np.array([5, 40, 7])

prediction = new_user @ beta

print("\ng. New user:")
print(new_user)

print("Predicted assistance score:")
print(prediction)
```
