```python
arr=np.array([1,2,3,4,5,1,2,1,1,1])
freq = {}
for x in arr:
    freq[x] = freq.get(x, 0) + 1
maxfreq = max(freq.values())
for x in freq:
    if freq[x] == maxfreq:
        indices = [i for i in range(len(arr)) if arr[i] == x]
        print("Value:", x)
        print("Frequency:", maxfreq)
        print("Indices:", indices)
```

```python
arr=np.array([1,1,1,2,3,4,2,4,3,3])
freq={}
for x in arr:
    freq[x]=freq.get(x,0)+1
maxfreq=max(freq.values())
for x in freq:
    if freq[x]==maxfreq:
       indices=[i for i in range(len(arr)) if arr[i]==x]
       print("value ",x)
       print("frequency ",maxfreq)
       print("indices ",indices)
```

```python
#q4
Vanshika=np.linspace(10,100,25)#generates an array of evenly spaced numbers over specified interval
print(Vanshika)
#dimensions
print(Vanshika.ndim)
#shape
print(Vanshika.shape)#for 2d it is row,col; for 3d it is depth,row,col
#total number of elements
print(Vanshika.size)
#data type of each element
print(Vanshika.dtype)
#total number of bytes consumed by the array
print(Vanshika.nbytes)
#reshape 
reshapedarr=Vanshika.reshape(Vanshika.size,1)
print(reshapedarr)
#reshape using .T attribute
Treshape=Vanshika.reshape(-1,1)#-1 is used to figure out the array's dimension by itself
arrreshape=Treshape.T
print(arrreshape)
```
